```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

def compare_population_dynamics(N0, K, Gs, delta, Ss, I_imm, time_steps=50):
    # Initialize arrays to store population over time
    N_capt = np.zeros(time_steps)
    N_Lcapt = np.zeros(time_steps)
    
    # Set initial populations
    N_capt[0] = N0
    N_Lcapt[0] = N0
    
    for t in range(1, time_steps):
        # 1. Original CAPTAIN
        # Geometric growth with mortality, capped at carrying capacity
        N_capt[t] = min(N_capt[t-1] * (1 - Ss * delta) * Gs + I_imm, K)
        
        # 2. Logistic CAPTAIN
        # Mortality applied first, followed by density-dependent logistic growth
        N_surv = max(N_Lcapt[t-1] * (1 - Ss * delta), 0)
        
        # Prevent division by zero if K is 0
        if K > 0:
            N_Lcapt[t] = ((K * N_surv * Gs) / (K + N_surv * (Gs - 1))) + I_imm
        else:
            N_Lcapt[t] = I_imm
            
    # Plotting the results
    plt.figure(figsize=(10, 6))
    
    # Plot Original CAPTAIN
    plt.plot(range(time_steps), N_capt, 
             label='Original CAPTAIN', 
             linestyle='--', color='red', linewidth=2)
    
    # Plot Logistic CAPTAIN
    plt.plot(range(time_steps), N_Lcapt, 
             label='Logistic CAPTAIN', 
             linestyle='-', color='blue', linewidth=2)
    
    # Plot Carrying Capacity Reference Line
    plt.axhline(K, color='black', linestyle=':', label='Carrying Capacity (K)')
    
    plt.xlabel('Time Step (t)')
    plt.ylabel('Population Size (N)')
    plt.title('Comparison of Population Dynamics: Original vs. Logistic CAPTAIN')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

# Generate interactive sliders
interact(compare_population_dynamics,
         N0=IntSlider(min=0, max=2000, step=10, value=100, description='Initial (N0)'),
         K=IntSlider(min=100, max=2000, step=50, value=1000, description='Capacity (K)'),
         Gs=FloatSlider(min=1.05, max=5.0, step=0.05, value=2.0, description='Growth (Gs)'),
         delta=FloatSlider(min=0, max=1.0, step=0.05, value=0.1, description='Disturbance'),
         Ss=FloatSlider(min=0, max=1.0, step=0.05, value=0.5, description='Sensitivity'),
         I_imm=IntSlider(min=0, max=100, step=1, value=2, description='Immigration'))
```
